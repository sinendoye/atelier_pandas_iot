## Partie 5: Manipulation des colonnes

### 1. Ajouter la colonne "temperature_fahrenheit"

```python
df["temperature_fahrenheit"] = df["temperature"] * 9/5 + 32
print("La nouvelle colonne temperature fahrenheit est:\n", df["temperature_fahrenheit"])
```

### 2. Ajouter la colonne "niveau_temperature" pour catégoriser la température en "Élevée" (température > 30) ou "Normale". On peut utiliser la méthode where() de numpy

```python
# np.where(condition, valeur_si_vrai, valeur_si_faux) fonctionne comme un if/else appliqué à toute la colonne d'un coup : pour chaque ligne, si temperature > 30, la nouvelle colonne vaut "Élevée", sinon "Normale"
df["niveau_temperature"] = np.where(df["temperature"] > 30, "Élevée", "Normale")
print("Les niveaux de temperature sont:\n",df["niveau_temperature"])
```

### 3. Renommer la colonne "humidite" en "humidite_relative"

```python
df.rename(columns={"humidite": "humidite_relative"}, inplace=True)
print(df["humidite_relative"])
```

### 4. Supprimer la colonne "temperature_fahrenheit"

```python
#inplace=True : par défaut, la plupart des méthodes Pandas ne modifient pas le DataFrame d'origine — elles en retournent une copie modifiée. inplace=True dit à Pandas de modifier df directement, sans avoir besoin de réécrire df = df.rename(...).
df.drop(columns=["temperature_fahrenheit"], inplace=True)
print(df)
```

## Partie 6: Filtrage

### 1. Récupérer uniquement les lignes telles que température > 30

```python
df_chaud = df[df["temperature"] > 30]
print("La table des températures supérieures à 30:\n", df_chaud)
```

### 2. Récupérer uniquement les lignes telles température > 30 et humidité > 70

```python
df_alerte = df[(df["temperature"] > 30) & (df["humidite_relative"] > 70)]
print("La table des alertes est:\n", df_alerte)
```

## Partie 9: Gestion des valeurs manquantes

### 1. Afficher le nombre de valeurs manquantes par colonne

```python
print("Le nombre de valeurs manquantes par colonne est de:\n", df.isnull().sum())
```

### 2. Afficher le nombre total des valeurs manquantes

```python
print("Le nombre total des valeurs manquantes est de: ", df.isnull().sum().sum())
```

### 3. Afficher le taux de valeurs manquantes

```python
taux_manquant = (df.isnull().sum().sum() / df.size) * 100
print(f"Le taux de valeurs manquantes est de : {taux_manquant:.2f} %")
```

### 4. Afficher les lignes contenant des valeurs manquantes

```python
print("Les lignes contenant des valeurs manquantes sont:\n", df[df.isnull().any(axis=1)])
```

### 5. Pour la température et l’humidité, remplacer les valeurs manquantes par les moyennes

```python
df.fillna({"temperature": df["temperature"].mean()}, inplace=True)
print("Temperature remplie \n", df["temperature"])

df.fillna({"humidite_relative": df["humidite_relative"].mean()}, inplace=True)
print("Humidite relative remplie \n", df["humidite_relative"])

#OU méthode aussi recommandée par la documentation Pandas
# df["temperature"] = df["temperature"].fillna(df["temperature"].mean())
# df["humidite_relative"] = df["humidite_relative"].fillna(df["humidite_relative"].mean())
```

### 6. Pour la consommation, remplacer les valeurs manquantes par la médiane

```python
df["consommation"]= df["consommation"].fillna(df["consommation"].median())
print("Consommation remplie: \n", df["consommation"])
```

### 7. Pour "etat", remplacer les valeurs manquantes par "INCONNU"

```python
df["etat"] = df["etat"].fillna("INCONNU")
print("Etat remplie:\n", df["etat"])
```
