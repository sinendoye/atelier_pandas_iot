```python
# PARTIE 1 : SERIES
# 1) Créer une Series contenant 4 températures
series_temp = pd.Series([24.5, 25.1, 26.3, 27.0])
print("1. La série de température : \n",series_temp)
# 2) Afficher son index
print("2. Index de la Series : ", series_temp.index)
# 3) Afficher la 1ère valeur (position 0)
print("3. Première valeur :", series_temp.iloc[0])
# 4) Afficher la dernière valeur (position -1)
print("4. Dernière valeur :", series_temp.iloc[-1])
# 5) Créer de deux façons différentes une Series avec index 12h, 13h, 14h, 15h
# Méthode 1 : Avec l'argument 'index'
series_heures_1 = pd.Series([24.5, 25.1, 26.3, 27.0], index=[12, 13, 14, 15])
print("Méthode 1 (index param) : \n",series_heures_1)
# Méthode 2 : Avec un dictionnaire (clé = index, valeur = donnée)
series_heures_2 = pd.Series({12: 24.5, 13: 25.1, 14: 26.3, 15: 27.0})
print("Méthode 2 (dictionnaire) : \n",series_heures_2)
# 6) Afficher la 1ère et la dernière valeur (en utilisant les labels d'index)
# Première valeur
print("Première valeur (à 12h) :", series_heures_1[12])
print("Dernière valeur (à 15h) :", series_heures_1[15])
```
